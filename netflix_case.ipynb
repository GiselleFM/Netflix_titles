1. Importar bibliotecas

```python
import pandas as pd
import numpy as np
```

2. Load o dataset

```python
df = pd.read_csv("C:/Users/gisel/Downloads/netflix_titles.csv/netflix_titles.csv")
```

3. Olhada inicial

```python
#Primeiras linhas do dataset
print('Primeiras linhas:')
display(df.head())

#Últimas linhas do dataset
print("\nÚltimas linhas:")
display(df.tail())

# Resumo do dataset
print("\nResumo:")
df.info()

# Número de linhas e colunas
print("\nShape do dataset:", df.shape)
```

4. Padronização dos dados

```python
#Conversão de tipo de dados
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')

#Separar duration em 2 colunas
df[['duration_value', 'duration_unit']] = df['duration'].str.split(' ', n=1, expand=True)

# Converter duration_value para númerico
df['duration_value'] = pd.to_numeric(df['duration_value'])

#Checar resultado
df[['duration', 'duration_value', 'duration_unit']].head()
```

```python
#Convertendo textos para minúsculos
colunas_texto = ['type', 'listed_in', 'rating', 'duration_unit']

#Aplicando
df[colunas_texto] = df[colunas_texto].apply(lambda col: col.str.lower())

#Verificar
df[['type', 'listed_in', 'rating', 'duration_unit']].head()
```

5. Limpeza dos dados

```python
#Valores nulos em director
print('Valores nulos em director:', df['director'].isna().sum())

#Valores nulos em cast
print('Valores nulos em cast:', df['cast'].isna().sum())

#Valores nulos em rating
print('Valores nulos em rating:', df['rating'].isna().sum())
```

```python
#Checando por dulplicates
print('Duplicados:', df.duplicated().sum())

#Preenchendo valores nulos em director, cast, rating
df.fillna({
    'director': 'unknow',
    'cast': 'unknow',
    'rating': 'not_rated'
}, inplace=True)

#Drop rows onde date_added é nulo por que precisamos de mês e ano para análise futura e so temos ano para preencher
df.dropna(subset=['date_added'], inplace=True)

#Verificando limpeza
df.info()
```

```python
#Verificando valores
df['rating'].unique()

#ERRO EM RATING(VALORES PERTECEM EM DURATION)
df[df['duration'].isna()]

# 1. Preencher duration com rating, mas só onde duration está nulo
df.loc[df["duration"].isna(), "duration"] = df["rating"]

# 2. Agora, limpar o rating apenas dessas linhas (onde houve a correção)
df.loc[df.index[df["rating"] == df["duration"]], "rating"] = None

#Verificando
df[df['duration'].isna()]
df['rating'].unique()
df.info()
```

```python
#Preencher country baseado em director

mapeamento = df.dropna(subset=["country"]).set_index("director")["country"].to_dict()
df["country"] = df["country"].fillna(df["director"].map(mapeamento))

#ainda existem valores nulos, podemos preencher a partir do cast

# Função para preencher country usando cast
def preencher_country_por_cast(df, cast_col="cast", country_col="country"):
    # Criar mapeamento de ator -> país com base em linhas que já têm country
    mapeamento_atores = {}
    for _, row in df.dropna(subset=[country_col]).iterrows():
        atores = [a.strip() for a in row[cast_col].split(",")]
        for ator in atores:
            mapeamento_atores[ator] = row[country_col]
    
    # Preencher country nas linhas nulas
    def buscar_pais(row):
        if pd.notna(row[country_col]):
            return row[country_col]
        atores = [a.strip() for a in row[cast_col].split(",")]
        for ator in atores:
            if ator in mapeamento_atores:
                return mapeamento_atores[ator]
        return np.nan
    
    df[country_col] = df.apply(buscar_pais, axis=1)
    return df

# Aplicar a função
df = preencher_country_por_cast(df)
print(df.info())

#Quase 200 registros sem paises, podemos dropar essas linhas
df.dropna(subset=['country'], inplace=True)
```

```python
# Salvar novo dataset
df.to_csv("C:/Users/gisel/Downloads/netflix_titles.csv/netflix_titles_clean.csv", index=False)
```
